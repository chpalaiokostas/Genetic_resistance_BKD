```julia
using CSV
using DataFrames
using DataFramesMeta
```

```julia
sire_assignment = CSV.read("Sire_assignment_results.txt",DataFrame)
select!(sire_assignment,[:Animal,:Sire])
```

```julia
unique(sire_assignment, :Animal)
```

```julia
@chain sire_assignment begin
    groupby(:Sire)
    combine(nrow => :family_size)
    sort(:family_size,rev=true)
end
```

```julia
dam_assignment = CSV.read("Dam_assignment_results.txt",DataFrame)
select!(dam_assignment,[:Animal,:Dam])
```

```julia
unique(dam_assignment, :Animal)
```

```julia
@chain dam_assignment begin
    groupby(:Dam)
    combine(nrow => :family_size)
    sort(:family_size,rev=true)
end
```

```julia
animals_both_parents = innerjoin(sire_assignment,dam_assignment,on=:Animal)
```

```julia
unique(animals_both_parents, :Animal)
```

```julia
@chain animals_both_parents begin
    groupby([:Sire,:Dam])
    combine(nrow => :family_size)
    sort(:family_size,rev=true)
end
```
